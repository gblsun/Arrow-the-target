```python
import numpy as np

import matplotlib.pyplot as plt
%matplotlib inline
```

```python
n_linhas = 100
n_colunas = 100

imagem01 = np.zeros((n_linhas, n_colunas, 3), dtype=np.uint8)

plt.figure(figsize=(3,3))
plt.imshow(imagem01);
```

```python
imagem02 = np.ones((n_linhas, n_colunas, 3), dtype=np.uint8)

plt.figure(figsize=(3,3))
plt.imshow(imagem02);
```

```python
imagem03 = 255 * np.ones((n_linhas, n_colunas, 3), dtype=np.uint8)

plt.figure(figsize=(3,3))
plt.imshow(imagem03);
```

```python
imagem04 = 255 * np.ones((n_linhas, n_colunas, 3), dtype=np.uint8)

imagem04[:,n_colunas//2,0] = 0
imagem04[:,n_colunas//2,1] = 0
imagem04[n_linhas//2,:,0] = 0
imagem04[n_linhas//2,:,1] = 0

plt.figure(figsize=(3,3))
plt.imshow(imagem04);
```

```python
imagem05 = 255 * np.ones((n_linhas, n_colunas, 3), dtype=np.uint8)

x, y = np.meshgrid(np.arange(n_colunas),
                   np.arange(n_linhas))
raio = 25

circulo = 255 * ((x-n_colunas//2)**2 + (y-n_linhas//2)**2 > raio**2).astype(np.uint8)

imagem05[:,:,0] = circulo
# imagem05[:,:,1] = circulo
# imagem05[:,:,2] = circulo

plt.figure(figsize=(3,3))
plt.imshow(imagem05);
```
